# Deskew one TIFF stack

This is the format-independent use of the package: load one stage-scan OPM volume with axes `(plane, y, x)`, deskew it, and save the Cartesian `(z, y, x)` result.

Time, channel, view, position and tile dimensions are outside the core function. If a dataset contains those dimensions, the surrounding workflow should split it into individual 3-D volumes and call `deskew_stage_scan` for each one.


```python
import tifffile

from opm_deskew import deskew_stage_scan
from opm_deskew.io import load_tiff_stack
```

```python
INPUT_TIFF = r"stage_scan.tif"
OUTPUT_TIFF = r"deskewed.tif"

stack = load_tiff_stack(INPUT_TIFF)
stack.shape
```

The expected input ordering is `(plane, camera_y, camera_x)`. A generic TIFF does not tell the package which physical direction increasing plane number represents, so the acquisition orientation must be known by the user or calling workflow.


```python
result = deskew_stage_scan(
    stack,
    pixel_size_um=0.35,
    scan_step_um=2.0,
    theta_deg=35.0,
    binning=1,
    reverse_plane_order=False,
    theta_interp_deg=None,
)

result.shape
```

```python
tifffile.imwrite(OUTPUT_TIFF, result)
```

If the source stack was stored in the opposite stage-scan order, set `reverse_plane_order=True`. `theta_interp_deg`, when supplied, changes only the in-plane sample locations used while interpolating between adjacent acquired planes; it does not change the acquisition-plane deskew geometry.
