# Micro-Manager stage-scan OPM workflow

This notebook runs the acquisition-specific workflow on a Micro-Manager folder. It uses the current dOPM folder naming and view-orientation conventions defined in `opm_deskew.workflows.micromanager`.

The core deskew routine itself does not know about Micro-Manager, views, time points or positions.


```python
from opm_deskew.workflows.micromanager import batch_process_micromanager
```

Set the acquisition and output folders plus the physical acquisition parameters. `theta_interp_deg` can be left as `None` for the default in-plane interpolation lookup.


```python
DATA_PATH = r"D:\Max\20260915_timelapse\data\y_stage_scan_20260915-183605"
OUT_PATH = r"D:\temp\Max\20260915_timelapse\data\y_stage_scan_20260915-183605"

pixel_size_um = 0.35
scan_step_um = 2.0
theta_deg = 35.0

binning = 4
workers = 12
```

```python
batch_process_micromanager(
    root_dir=DATA_PATH,
    output_dir=OUT_PATH,
    pixel_size_um=pixel_size_um,
    scan_step_um=scan_step_um,
    theta_deg=theta_deg,
    binning=binning,
    workers=workers,
    theta_interp_deg=None,
    live_mode=True,
    resume=True,
    save_mip=True,
)
```

The workflow finds the current `Hamamatsu` camera folders, loads each 3-D volume, applies the system-specific acquisition ordering/orientation rules, deskews it, and writes the reconstructed volume and optional MIP.

For a different Micro-Manager folder layout or a different microscope orientation, adapt the workflow rather than changing the core deskew geometry.
